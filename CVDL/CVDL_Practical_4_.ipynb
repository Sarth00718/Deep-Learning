# Liner Filters

Purpose: Smooth, sharpen, or detect edges in an image using linear operations.
Each output pixel is a weighted sum of its neighbors.

Examples:

- Averaging filter (Mean filter): Reduces noise by averaging neighboring pixels.

- Sharpening (Convolution with kernel): Enhances edges and details.

## Average Filtering

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image = cv2.imread("cameramen.png", 0)

rows, cols = image.shape
output = np.zeros((rows, cols), dtype=np.uint8)

mask = np.ones((3, 3), np.float32) / 9

for i in range(1, rows-1):
    for j in range(1, cols-1):
        region = image[i-1:i+2, j-1:j+2]
        output[i, j] = np.sum(region * mask)

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(image, cmap='gray')
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(output, cmap='gray')
plt.title("Averaging Filtered")
plt.axis("off")

plt.show()
```

##Convolution Filtering

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image = cv2.imread("cameramen.png", 0)

rows, cols = image.shape
output = np.zeros((rows, cols), dtype=np.uint8)

# (Sharpen)
mask = np.array([[0, -1, 0],
                 [-1, 5, -1],
                 [0, -1, 0]], dtype=np.float32)

for i in range(1, rows-1):
    for j in range(1, cols-1):
        region = image[i-1:i+2, j-1:j+2]
        value = np.sum(region * mask)
        output[i, j] = np.clip(value, 0, 255)

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(image, cmap='gray')
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(output, cmap='gray')
plt.title("Convolution Filtered")
plt.axis("off")

plt.show()
```

# Non Liner Filters

Median filter is best for removing salt-and-pepper noise.

Maximum filter brightens objects, removes pepper noise.

Minimum filter darkens objects, removes salt noise.

#Edge Detection Filters

- Sobel/Prewitt: Good for clear horizontal/vertical edges; Sobel is more sensitive due to weighting.

- Roberts: Detects fine diagonal edges; may be noisier.

- Laplacian: Captures edges in all directions but sensitive to noise; often used with smoothing first.

##Sobel Filtering

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image = cv2.imread("bplane.jpg", 0)
sobel_x = np.array([[-1, 0, 1],
                    [-2, 0, 2],
                    [-1, 0, 1]])
sobel_y = np.array([[-1, -2, -1],
                    [0,  0,  0],
                    [1,  2,  1]])
rows, cols = image.shape
output_x = np.zeros((rows, cols), dtype=np.float32)
output_y = np.zeros((rows, cols), dtype=np.float32)

for i in range(1, rows-1):
    for j in range(1, cols-1):
        region = image[i-1:i+2, j-1:j+2]
        output_x[i, j] = np.sum(region * sobel_x)
        output_y[i, j] = np.sum(region * sobel_y)

magnitude = np.sqrt(output_x**2 + output_y**2)
magnitude = np.clip(magnitude, 0, 255).astype(np.uint8)

plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(image, cmap='gray')
plt.title("Original (eight.tif)")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(magnitude, cmap='gray')
plt.title("Sobel Magnitude")
plt.axis("off")

plt.show()
```

##Robert's Filtering

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image = cv2.imread("lenaimg.png", 0)

roberts_x = np.array([[1, 0],
                      [0,-1]])
roberts_y = np.array([[0, 1],
                      [-1, 0]])

rows, cols = image.shape
gx = np.zeros((rows-1, cols-1), dtype=np.float32)
gy = np.zeros((rows-1, cols-1), dtype=np.float32)

for i in range(rows-1):
    for j in range(cols-1):
        region = image[i:i+2, j:j+2]
        gx[i, j] = np.sum(region * roberts_x)
        gy[i, j] = np.sum(region * roberts_y)

magnitude = np.sqrt(gx**2 + gy**2)
magnitude = np.clip(magnitude, 0, 255).astype(np.uint8)

plt.figure(figsize=(8, 8))
plt.subplot(1, 2, 1)
plt.imshow(image, cmap='gray')
plt.title("Original")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(magnitude, cmap='gray')
plt.title("Roberts Magnitude")
plt.axis("off")

plt.show()
```
