# Image Negative Transformation

Image negative transformation is a simple image processing operation that inverts the brightness and colors of an image. Dark areas become bright, and bright areas become dark.

## Formula

\[
s = L - 1 - r
\]

## Use Cases

- **Medical Imaging:** Makes features in dark regions more visible.
- **Photography:** Produces photographic negatives used in analog printing.


```python
negative_image = 255 - grayscale
sbs = cv2.hconcat([grayscale, negative_image])

print("Grayscale and Negative Image:")
cv2_imshow(sbs)

color_negative = 255 - image
sbs_color = cv2.hconcat([image, color_negative])

print("Original and Color Negative Image:")
cv2_imshow(sbs_color)
```

# Log Transformation

Log transformation in image processing is a technique that applies a logarithmic function to each pixel value in an image. This operation is particularly useful for expanding the range of dark pixels while compressing the range of bright pixels.

## Formula
\[
s = c . log(1 + r)
\]

Where:
- \( c \) is a scaling constant
  \[
  c = 255 / log(1 + R_max)
  \]


## Use Cases

- **Highlighting details in dark regions**
- **Dynamic range compression**

```python
import matplotlib.pyplot as plt

image = cv2.imread("log_transform.jpg", 0).astype(np.float32)

c = 255 / np.log(1 + np.max(image))
log_transformed = c * np.log(1 + image)
log_transformed = np.array(log_transformed, dtype=np.uint8)

image_uint8 = np.array(image, dtype=np.uint8)

sbs = cv2.hconcat([image_uint8, log_transformed])
print("Original and Log Transformed Image:")
cv2_imshow(sbs)

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.hist(image_uint8.ravel(), bins=256, range=(0, 256))
plt.title('Histogram of Original Image')
plt.xlabel('Pixel Value')
plt.ylabel('Frequency')

plt.subplot(1, 2, 2)
plt.hist(log_transformed.ravel(), bins=256, range=(0, 256))
plt.title('Histogram of Log Transformed Image')
plt.xlabel('Pixel Value')
plt.ylabel('Frequency')

plt.tight_layout()
plt.show()
```

# Power Law (Gamma) Transformation

Power law (also called gamma) transformation is a point operation in image processing that adjusts the intensity of an image using an exponential function. This method allows flexible enhancement of either dark or bright regions, depending on the gamma value.

## Formula

\[
s = c . r^{gamma}
\]

Where:
  - \( gamma < 1 \) : Expands dark areas, brightens the image.
  - \( gamma > 1 \) : Compresses bright areas, darkens the image.

## Use Cases

- **Enhancing details** in specific regions of images
- **Medical imaging and printing**

```python


image = cv2.imread("Grayscale.jpeg", 0).astype(np.float32)

gamma = 1.5
c = 255 / (np.max(image) ** gamma)
power_law_transformed = c * (image ** gamma)
power_law_transformed = np.array(power_law_transformed, dtype=np.uint8)

image_uint8 = np.array(image, dtype=np.uint8)

sbs = cv2.hconcat([image_uint8, power_law_transformed])
print("Original and Power-Law Transformed Image:")
cv2_imshow(sbs)
```

# Contrast Stretching

Contrast stretching is a fundamental image enhancement technique used to improve the visual quality of an image by expanding the range of pixel intensity values. It helps to increase the contrast of an image by stretching the original narrow range of intensity values to a wider range, often spanning the full available intensity scale.

## Formula

\[
I = {I - I_min} * (L - 1) / {I_max - I_min}
\]


## Use Cases

- Enhancing details in medical images (X-rays, MRI)
- Improving the visibility in surveillance footage
- Enhancing photos to be more visually appealing
- Satellite and aerial image analysis (remote sensing)

```python

image = cv2.imread("input.tif", 0).astype(np.float32)

min_val = np.min(image)
max_val = np.max(image)
stretched = 255 * (image - min_val) / (max_val - min_val)
stretched = np.array(stretched, dtype=np.uint8)

image_uint8 = np.array(image, dtype=np.uint8)

sbs = cv2.hconcat([image_uint8, stretched])
print("Original and Contrast Stretched Image:")
cv2_imshow(sbs)
```
